# 03 Figures — phase diagram + tradeoffs + time series

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import pandas as pd, matplotlib.pyplot as plt
from utils.plotting import phase_heatmap, tradeoff_curve
from src.simulation import SimParams, run_coupled

df = pd.read_csv(ROOT / 'data' / 'pilot_full.csv')
fig, axes = plt.subplots(1,2, figsize=(10,4))
phase_heatmap(df, 'final_size', 'observation', 'ws', ax=axes[0])
tradeoff_curve(df, ax=axes[1])
plt.tight_layout(); plt.show()

# example dynamics: 1-wave vs multi-wave
for d in [0.0, 0.1, 0.6]:
    out = run_coupled(SimParams(topology='ws', n=500, seed=0, rule='observation', alpha=0.8, delta=d))
    plt.plot(out['prev_series'], label=f'delta={d} waves={out["waves"]}')
plt.legend(); plt.xlabel('step'); plt.ylabel('prevalence'); plt.show()
```
